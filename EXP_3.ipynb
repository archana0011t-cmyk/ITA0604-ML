```python
from sklearn.tree import DecisionTreeClassifier

X = [[0,0], [0,1], [1,0], [1,1]]
y = ['No', 'Yes', 'Yes', 'Yes']

model = DecisionTreeClassifier(criterion='entropy')
model.fit(X, y)

sample = [[1, 0]]
print("Prediction:", model.predict(sample)[0])
```
