```python
from sklearn.linear_model import LinearRegression

X = [
    [1, 10000],
    [2, 20000],
    [3, 30000],
    [4, 40000],
    [5, 50000]
]

y = [900000, 800000, 700000, 600000, 500000]

model = LinearRegression()
model.fit(X, y)

print("Predicted Price:", model.predict([[2, 15000]])[0])
```
