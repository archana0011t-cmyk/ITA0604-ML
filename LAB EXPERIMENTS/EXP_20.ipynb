```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5], [6]]
y = [100, 120, 140, 160, 180, 200]

model = LinearRegression()
model.fit(X, y)

print("Predicted Sales:", model.predict([[7]])[0])
```
