```python
from sklearn.linear_model import LinearRegression

X = [[500], [800], [1000], [1200], [1500]]
y = [20, 32, 40, 48, 60]

model = LinearRegression()
model.fit(X, y)

print("Predicted Price:", model.predict([[1100]])[0], "Lakhs")
```
