```python
from sklearn.tree import DecisionTreeClassifier

X = [[300], [450], [550], [650], [750], [850]]
y = ['Poor', 'Poor', 'Average', 'Good', 'Good', 'Excellent']

model = DecisionTreeClassifier()
model.fit(X, y)

print("Credit Class:", model.predict([[700]])[0])
```
