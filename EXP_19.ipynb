```python
from sklearn.naive_bayes import GaussianNB

X = [
    [20000, 500],
    [25000, 550],
    [60000, 700],
    [80000, 750],
    [90000, 800]
]

y = ['No', 'No', 'Yes', 'Yes', 'Yes']

model = GaussianNB()
model.fit(X, y)

print("Loan Prediction:", model.predict([[65000, 720]])[0])
```
