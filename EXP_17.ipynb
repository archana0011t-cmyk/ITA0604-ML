```python
from sklearn.tree import DecisionTreeClassifier

X = [
    [2, 3000],
    [3, 4000],
    [4, 6000],
    [6, 8000],
    [8, 12000]
]

y = [0, 1, 1, 2, 3]

model = DecisionTreeClassifier()
model.fit(X, y)

print("Price Range:", model.predict([[6, 7000]])[0])
```
