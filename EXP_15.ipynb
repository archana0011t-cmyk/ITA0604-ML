```python
from sklearn.datasets import load_iris
from sklearn.naive_bayes import GaussianNB

iris = load_iris()

X = iris.data
y = iris.target

model = GaussianNB()
model.fit(X, y)

p = model.predict([X[0]])

print("Prediction:", iris.target_names[p[0]])
print("Accuracy:", model.score(X, y))
```
